```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("DataFramesDemo") \
    .master("local[*]") \
    .getOrCreate()

print("Spark Session Created Successfully!")

print("Spark Version:", spark.version)
```

```python
# Sample employee data
data = [
    (101, "Amit", 25, 25000),
    (102, "Neha", 30, 35000),
    (103, "Rahul", 28, 30000),
    (104, "Priya", 32, 40000),
    (105, "Rohan", 27, 28000)
]

columns = ["EmployeeID", "Name", "Age", "Salary"]

df = spark.createDataFrame(data, columns)
```

```python
print("Employee Data:")
df.show()

# Generate statistical summary
print("Statistical Summary:")
df.describe(["Age", "Salary"]).show()
```

```python
# Detailed statistical summary
df.select("Age", "Salary").summary().show()
```

```python
from pyspark.sql.functions import (
    count,
    mean,
    min,
    max,
    stddev,
    variance,
    sum as spark_sum
)
df.select(
    count("Salary").alias("Count"),
    spark_sum("Salary").alias("TotalSalary"),
    mean("Salary").alias("AverageSalary"),
    min("Salary").alias("MinimumSalary"),
    max("Salary").alias("MaximumSalary"),
    stddev("Salary").alias("StandardDeviation"),
    variance("Salary").alias("Variance")
).show(truncate=False)
```
