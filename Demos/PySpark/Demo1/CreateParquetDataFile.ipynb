```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("DataFramesDemo") \
    .master("local[*]") \
    .getOrCreate()

print("Spark Session Created Successfully!")

print("Spark Version:", spark.version)
```

```python
data = [
    (101, "Amit", 25, "Pune", 25000),
    (102, "Neha", 30, "Mumbai", 35000),
    (103, "Rahul", 28, "Delhi", 30000),
    (104, "Priya", 32, "Pune", 40000)
]

columns = [
    "EmployeeID",
    "Name",
    "Age",
    "City",
    "Salary"
]

df = spark.createDataFrame(data, columns)
df.show()
df.write.mode("overwrite").parquet("employee_parquet")
```
