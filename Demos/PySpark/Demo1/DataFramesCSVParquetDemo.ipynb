```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("DataFramesDemo") \
    .master("local[*]") \
    .getOrCreate()

print("Spark Session Created Successfully!")

print("Spark Version:", spark.version)
```

```python
# Read CSV
df = spark.read.csv(
    "employees.csv",
    header=True,
    inferSchema=True
)

print("CSV Data:")
df.show()

# Save as Parquet
df.write \
    .mode("overwrite") \
    .parquet("employee_parquet")

# Read Parquet
parquet_df = spark.read.parquet("employee_parquet")

print("Parquet Data:")
parquet_df.show()

parquet_df.printSchema()
```
